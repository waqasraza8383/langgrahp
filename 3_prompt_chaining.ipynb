```python
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph
from langgraph.graph import START
from langgraph.graph import END
from typing import TypedDict
from dotenv import load_dotenv

```

```python
load_dotenv()
```

```python
model =ChatOpenAI()
```

```python
class BlogeState(TypedDict):
    title: str
    outline:str
    content:str


```

```python
graph=StateGraph(BlogeState)

```

```python
def create_outline(state:BlogeState)->BlogeState:
    #fetch title
    title=state['title']
    #generate a prompt
    prompt=f"generate a detail bloc on the {title}"
    #update the state
    outline=model.invoke(prompt).content
    #update the state 
    state['outline']=outline

    return state
```

```python
def create_blog(state:BlogeState)->BlogeState:
    title=state['title']
    outline=state['outline']
    #now generate a prompt 
    prompt=f"write a detail not on the {title}  on the {outline}"
    content=model.invoke(prompt).content
    state['content']=content
    
```

```python
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)


graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_edge')
graph.add_edge('create_edge',END)


workflow=graph.compile()
```

```python
inital_state={'title':"tell me about the pakistan"}
final_state=workflow.invoke(inital_state)

print(final_state)


```
